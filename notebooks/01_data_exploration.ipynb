```python
import pandas as pd
data = pd.read_csv("../data/processed/clean_passwords.csv")
data['password'] = data['password'].astype(str)
data['length'] = data['password'].apply(len)
data['lowercase_count'] = data['password'].apply(lambda x: sum(1 for c in x if c.islower()))
data['uppercase_count'] = data['password'].apply(lambda x: sum(1 for c in x if c.isupper()))
data['digit_count'] = data['password'].apply(lambda x: sum(1 for c in x if c.isdigit()))
data['special_count'] = data['length'] - (data['lowercase_count'] + data['uppercase_count'] + data['digit_count'])
data.head()
```

```python
import math
from collections import Counter

def calculate_entropy(password):
    if not password:
        return 0
    entropy = 0
    length = len(password)
    counts = Counter(password)
    for count in counts.values():
        probability = count / length
        entropy -= probability * math.log2(probability)
        return entropy

data['entropy'] = data['password'].apply(calculate_entropy)
data.to_csv("../data/processed/engineered_passwords.csv", index=False)
data.head()
```
